```python
import pandas as pd

# データフレーム作成
data = {
    "日付": ["2023-05-17", "2023-05-18", "2023-05-19", "2023-05-20", "2023-05-21"],
    "社員名": ["山田", "佐藤", "鈴木", "田中", "高橋"],
    "売上": [100, 200, 150, 300, 250],
    "部門": ["メーカー", "代理店", "メーカー", "商社", "代理店"]
}

df = pd.DataFrame(data)

# 平均売上を算出
average_sales = df["売上"].mean()

# 平均売上列を追加
df["平均売上"] = average_sales

# 業績ランクを追加
def rank(sales):
    if sales >= 250:
        return "A"
    elif sales >= 200:
        return "B"
    else:
        return "C"

df["業績ランク"] = df["売上"].apply(rank)

# Excelファイルへ書き出し
output_file = "sales_report.xlsx"
df.to_excel(output_file, index=False)

# 結果表示
print(df)
print(f"\n平均売上: {average_sales}")

print(f"\nExcelファイル '{output_file}' を作成しました。")
```
